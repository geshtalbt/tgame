# Задание 1. Камень, Ножницы, Бумага (на победу)
### Суть: Классическая игра против компьютера, но с условием: играть нужно до 3-х побед.
Основные требования:
1. Компьютер случайным образом выбирает один из трех вариантов.
2. Игрок вводит свой вариант
3. Вывод результата раунда (Победа / Поражение / Ничья) и общего счета.
4. Игра идет в цикле, пока кто-то не наберет 3 очка.
5. В конце предложение сыграть еще раз.


```python
import random

variants = ['paper', 'rock', 'scissors']

print('''-----------------------------
|Paper, Rock, Scissors game'|
-----------------------------''')

while True:

  comp_score = 0
  user_score = 0

  while comp_score < 3 and user_score <3:
    comp = random.choice(variants)
    user = input('Enter your choise (paper,rock or scissors): ')

    if user == comp:
      print(f'you both choise {user}, draw')
    elif (user == 'rock' and comp == 'scissors') or (user == 'paper' and comp == 'rock'):
      user_score += 1
      print(f'Computer choise {comp}, you win!')
    else:
      comp_score += 1
      print(f'Computer choise {comp},you lose')

    print(f'''Computer score = {comp_score}
Your score = {user_score}''')

  if user_score == 3:
    print('You won the game!')
  else:
    print('Computer won the game')

  again = input("Play again? (yes/no): ")
  if again != 'yes':
      break
```

# Задание 2. Виселица (Угадай слово)
### Суть: Игрок угадывает скрытое слово, называя буквы. За каждую ошибку отнимается «жизнь».
Основные требования:
1. Заранее подготовленный массив/список слов. Случайный выбор одного.
2. Слово отображается в виде прочерков или нижних подчеркиваний (например: _ _ _ _ _).
3. При вводе буквы программа проверяет, есть ли она в слове. Если да — открывает все такие буквы. Если нет — вычитает жизнь.
4. Вывод сообщения о победе или поражении.


```python
# Ваше решение здесь
import random

print('''--------------
|Hangman game|
--------------''')

variants = ['robber', 'plane', 'bank', 'car', 'watermelon', 'catterpilar']
word = random.choice(variants)
guessed = ['_' for _ in word]
lives = 6
used = []

while True:

  while lives >0 and '_' in guessed:
    print(' '.join(guessed))
    letter = input('Enter a letter: ')
    if letter in used:
      print('Letter have alredy used')
      continue

    used.append(letter)

    if letter in word:
      for i in range(len(word)):
        if word[i] == letter:
          guessed[i] = letter

    else:
      lives -= 1
      print(f'Wrong letter, lives left: {lives}')

  if '_' != guessed:
    print(f'You Won! The word was: {word}')
  else:
    print(f'You lost! The word was: {word}')

  again = input("Play again? (yes/no): ")
  if again != 'yes':
      break
```

# Задание 3. Угадай число
### Суть: Программа загадывает число от 1 до 100. Пользователь пытается угадать это число.
Основные требования:
1. После попытки угадать программа сообщает один из вариантов:
  - Не угадал! Загаданное число БОЛЬШЕ
  - Не угадал! Загаданное число МЕНЬШЕ
  - Поздравляем ты угадал число!
2. Программа считает количество неудачных попыток и выводит счет в конце игры.
3. \* Улучшите игру так чтобы пользователь мог сам указывать диапазон загаднного числа. Диапазон не должен быть меньше 10. И не должен быть отрицательным.
4. \** Определить минимальное необходимое число попыток для нахождения загаданного числа. Если пользователь не укладывается в это количество попыток - то он проигрывает. (Задача учитывается только при выполнении пункта 3.

```python
# Ваше решение здесь
import random

print('''------------------------
|Guesse the number game|
------------------------''')

min = (int(input('Enter min number: ')))
max = (int(input('Enten max number: ')))

while max - min <10 or min <0:
  print('Range must be at least 10 and not negative, try again')
  min = (int(input('Enter min number: ')))
  max = (int(input('Enten max number: ')))

number = random.randint(min, max)
attemps = 1

diapazon = max - min + 1
max_attemps = 0
while diapazon > 1:
  diapazon = diapazon // 2
  max_attemps += 1

print(f'You have {max_attemps} attemps to guesse number')

while True:

  user_number = int(input(f'Enter a number from {min} to {max}: '))

  if user_number < number:
    print(f'A number greater than {user_number}, try again')
    attemps += 1
  elif user_number > number:
    print(f'A number lower than {user_number}, try again')
    attemps += 1
  else:
    print(f'Congratulations you guessed the number! Attemps: {attemps}')
    break
  if attemps > max_attemps:
    print(f'The number was {number}, you lose!')
    break
```

```python
x = np.linspace(0, 10, 300)
y1 = np.sin(x) * np.exp(-x / 10)
y2 = np.cos(x) * np.exp(-x / 10)

fig, ax = plt.subplots(figsize=(10, 5), dpi=120)

ax.plot(x, y1, color='#FF6B6B', linewidth=2.5, label='Затухающий синус')
ax.plot(x, y2, color='#4ECDC4', linewidth=2.5, label='Затухающий косинус')

ax.fill_between(x, y1, alpha=0.3, color='#FF6B6B')
ax.fill_between(x, y2, alpha=0.3, color='#4ECDC4')

ax.set_title('Затухающие колебания', fontsize=16, fontweight='bold', pad=15)
ax.set_xlabel('Время', fontsize=12)
ax.set_ylabel('Амплитуда', fontsize=12)
ax.legend(fontsize=11, loc='upper right')
ax.grid(True, alpha=0.3)

plt.show()
```
